```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df=pd.read_csv('co2.csv')
df=df.drop_duplicates(ignore_index=True)
```

```python
y=df['CO2 Emissions(g/km)'].values
x=df['Fuel Consumption Comb (L/100 km)'].values
plt.scatter(x,y)
plt.title('CO2 Emissions vs Fuel Consumption')
plt.xlabel('Fuel Consumption Comb (L/100 km)')   
plt.ylabel('CO2 Emissions(g/km)')
plt.show()
```

```python
def MSE(x,y,w,b):
    n=len(x)
    sum=0
    for i in range(n):
        y_pred=w*x[i]+b
        error=(y[i]-y_pred)**2
        sum+=error
    MSE=1/(2*n)*sum
    return MSE
```

```python
def gradient(x,y,w,b):
    n=len(x)
    dw=0
    db=0
    for i in range(n):
        y_pred=w*x[i]+b
        dw+=x[i]*(y_pred-y[i])
        db+=(y_pred-y[i])
    dw=dw/n
    db=db/n
    return dw,db
```

```python
def gradient_descent(x,y,A,its):
    w=0
    b=0
    for i in range(its):
        dw,db=gradient(x,y,w,b)
        w=w-A*dw
        b=b-A*db
        if i%1000==0:
            print(f"Iteration {i}: MSE={MSE(x,y,w,b)}")
    return w,b

    
```

```python
A=0.001
its=10000
df=df.drop_duplicates(ignore_index=True)
y=df['CO2 Emissions(g/km)'].values
x=df['Fuel Consumption Comb (L/100 km)'].values
fw,fb=gradient_descent(x,y,A,its)
plt.scatter(x,y)
x1=np.linspace(min(x),max(x),100)
y1=fw*x1+fb
plt.plot(x1,y1,color='red',label='Regression Line')
plt.title('CO2 Emissions vs Fuel Consumption')
plt.xlabel('Fuel Consumption Comb (L/100 km)')   
plt.ylabel('CO2 Emissions(g/km)')
plt.legend()
plt.show()
```
