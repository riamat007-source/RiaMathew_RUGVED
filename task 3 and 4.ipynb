```python
import matplotlib.pyplot as plt
import math
position=[0,0,0]
n=1
waypoints=[(0,0),]
def pos(move):
  if move[0].lower()=="forward":
    position[1]+=int(move[1])
  elif move[0].lower()=="left":
    position[0]-=int(move[1])
  elif move[0].lower()=="right":
    position[0]+=int(move[1])
  if position[0]!=0 and position[1]!=0:
      theta=math.degrees(math.atan2(position[1],position[0]))
      position[2]=theta
  waypoints.append((position[0],position[1]))
  return position
while n==1:
  m=input("Enter movement in the form forward <val>,left <val> and right <val>")
  move=m.split()
  print("initial position",position)
  print("executing ",m)
  print("final position",pos(move))
  n=int(input("Enter 1 to continue and 0 to exit"))
x=[]
y=[]
for i in waypoints:
    x.append(i[0])
    y.append(i[1])
plt.plot(x,y)
plt.scatter(x[0], y[0], marker='s', label='Start')
plt.scatter(x[-1], y[-1], marker='X', label='End')
for i in range(len(x) - 1):
    dx=x[i+1]-x[i]
    dy=y[i+1]-y[i]
    plt.quiver(x[i], y[i],dx,dy)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Trajectory")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import math
x=0
y=0
theta=0
dt=0.01
position=[0,0,0]
n=1
waypoints=[(0,0),]
def pos(move):
  global x,y,dt,position,theta
  steps=move[2]/dt
  for i in range(int(steps)):
     x=x+move[0]*math.cos(theta)*dt
     y=y+move[0]*math.sin(theta)*dt
     theta=theta+move[1]*dt
     waypoints.append((x,y))
  position[0]=x
  position[1]=y
  position[2]=theta
  return position
while n==1:
  move=eval(input("Enter movement in the form [velocity,angular velocity,time]"))
  print("initial position",position)
  print("executing ",move)
  print("final position",pos(move))
  n=int(input("Enter 1 to continue and 0 to exit"))
x=[]
y=[]
for i in waypoints:
    x.append(i[0])
    y.append(i[1])
plt.plot(x,y)
plt.scatter(x[0], y[0], marker='s', label='Start')
plt.scatter(x[-1], y[-1], marker='X', label='End')
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Trajectory")
plt.grid(True)
plt.show()
```
